```python
# Loading Data
import streamlit as st
import pandas as pd

df = pd.read_csv("/Users/anamleen/Desktop/Travel details dataset.csv")

print(df.head())
```

```python
# Drop rows with missing values
df = df.dropna()

print(df.head())
```

```python
# Drop duplicate values
df = df.drop_duplicates()
initial_data = len(df)
print("Removed", initial_data - len(df), "duplicate rows")
```

```python
# Convert Flight and Airplane to Plane in the Transportation Type column
df["Transportation type"] = df["Transportation type"].replace({
    "Airplane": "Plane",
    "Flight": "Plane"
})

print(df.head())
```

```python
# Checking if data types are in the right form
df.info()
```

```python
# Converting Start and End date into datetime 
df["Start date"] = pd.to_datetime(df["Start date"], format="%m/%d/%Y")
df["End date"] = pd.to_datetime(df["End date"], format="%m/%d/%Y")

df.info()
```

```python
# Converting cost columns into integers
for col in ["Accommodation cost", "Transportation cost"]:
    df[col] = df[col].astype(str).str.replace(r"[^0-9.]", "", regex=True)
    df[col] = pd.to_numeric(df[col])

print(df.head())
```
